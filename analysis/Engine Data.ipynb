```python
import pandas as pd

df = pd.read_csv("train_FD001.txt", sep=r"\s+", header=None)

df.head()
```

```python
import matplotlib.pyplot as plt

engine = df[df["unit_number"] == 1]

plt.figure(figsize=(12, 6))
plt.plot(engine["time_in_cycles"], engine["sensor_4"])
plt.xlabel("Operating Cycles")
plt.ylabel("Sensor 4")
plt.title("Sensor 4 Trend - Engine 1")
plt.show()
```

```python
sensors = ["sensor_2", "sensor_3", "sensor_4", "sensor_7", "sensor_11", "sensor_12", "sensor_15"]

engine = df[df["unit_number"] == 1]

for sensor in sensors:
  plt.figure(figsize=(10, 4))
  plt.plot(engine["time_in_cycles"], engine[sensor])
  plt.xlabel("Operating Cycles")
  plt.ylabel(sensor)
  plt.title(f"{sensor} Trend - Engine 1")
  plt.show()
```

```python
sensor_cols = [f"sensor_{i}" for i in range(1, 22)]

correlations = df[sensor_cols + ["time_in_cycles"]].corr()["time_in_cycles"]
correlations = correlations.drop("time_in_cycles").sort_values()

correlations
```

```python
life = df.groupby("unit_number")["time_in_cycles"].max().sort_values()

shortest = life.head(5)
longest = life.tail(5)

representative = pd.concat([shortest, longest])

plt.figure(figsize=(10, 6))
plt.barh(
    representative.index.astype(str),
    representative.values
)

plt.xlabel("Useful Life (Cycles)")
plt.ylabel("Engine")
plt.title("Shortest and Longest Engine Useful Lives")
plt.tight_layout()
plt.show()
```

```python
# Calculate remaining useful life (RUL for every observation)
df["RUL"] = df.groupby("unit_number")["time_in_cycles"].transform("max") - df["time_in_cycles"]

df[["unit_number", "time_in_cycles", "RUL"]].head()
```

```python
df.corr(numeric_only=True)["RUL"].sort_values()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["RUL"], df["sensor_11"], alpha=0.4)

plt.xlabel("Remaining Useful Life (Cycles)")
plt.ylabel("Sensor 11")
plt.title("Sensor 11 vs. Remaining Useful Life")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = df["RUL"]
y = df["sensor_11"]

plt.figure(figsize=(10, 6))

plt.scatter(x, y, alpha=0.3)

# Add trendline
z = np.polyfit(x, y, 1)
p = np.poly1d(z)
plt.plot(x, p(x), linewidth=2)

plt.xlabel("Remaining Useful Life (Cycles)")
plt.ylabel("Sens")
```
